```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data.csv")
df
```

```python
X = df['Area']
y = df['Price']

X = X / 1000

n = len(X)
```

```python
m, c = 0, 0 
lr = 0.001 
epochs = 20000
loss_history = []
```

```python
for i in range(epochs):
    y_pred = m * X + c

    error = y_pred - y

    loss = 1/n * np.sum(error ** 2)
    loss_history.append(loss)

    dm = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)

    m = m - lr * dm
    c = c - lr * dc
print('Final parameters -> m:', m, ' c:', c)    
loss_history
    
```
